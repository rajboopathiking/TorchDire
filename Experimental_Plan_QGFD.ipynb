```python
from transformers import AutoModelForCausalLM, AutoTokenizer, BitsAndBytesConfig
from peft import LoraConfig, prepare_model_for_kbit_training
from trl import SFTTrainer, SFTConfig
from datasets import load_dataset

model_name = "EleutherAI/gpt-neo-2.7B"

# 4-bit quantization config (QLoRA) — drop this and load in fp16/bf16 if you have enough VRAM
bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_use_double_quant=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype="bfloat16",
)

model = AutoModelForCausalLM.from_pretrained(
    model_name,
    quantization_config=bnb_config,
    device_map="auto",
)



model = prepare_model_for_kbit_training(model)

tokenizer = AutoTokenizer.from_pretrained(model_name)
if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

# LoRA config — GPT-Neo attention modules are named differently from Llama-style models
lora_config = LoraConfig(
    r=16,
    lora_alpha=32,
    lora_dropout=0.05,
    bias="none",
    task_type="CAUSAL_LM",
    target_modules=["k_proj", "v_proj", "q_proj", "out_proj"],  # GPT-Neo attention proj names
)

# --- Dataset ---
ds = load_dataset("arbml/alpagasus_cleaned")

# Check the actual columns first — alpaca-style sets usually have
# instruction / input / output. Adjust the formatting function to match.
print(ds["train"].column_names)
print(ds["train"][0])

def format_example(example):
    if example.get("input"):
        prompt = (
            f"### Instruction:\n{example['instruction']}\n\n"
            f"### Input:\n{example['input']}\n\n### Response:\n{example['output']}"
        )
    else:
        prompt = (
            f"### Instruction:\n{example['instruction']}\n\n### Response:\n{example['output']}"
        )
    return {"text": prompt}

train_ds = ds["train"].map(format_example)

# --- Training ---
training_args = SFTConfig(
    output_dir="./gpt-neo-2.7b-alpagasus-lora",
    per_device_train_batch_size=4,
    gradient_accumulation_steps=4,
    num_train_epochs=3,
    learning_rate=2e-4,
    bf16=True,
    logging_steps=10,
    save_strategy="epoch",
    dataset_text_field="text",
    packing=False,
)

trainer = SFTTrainer(
    model=model,
    train_dataset=train_ds,
    args=training_args,
    peft_config=lora_config
)

trainer.train()
trainer.save_model("./gpt-neo-2.7b-alpagasus-lora-final")
```

```python


from transformers import AutoModelForCausalLM, AutoTokenizer, BitsAndBytesConfig
from peft import LoraConfig, prepare_model_for_kbit_training
from trl import SFTTrainer, SFTConfig
from datasets import load_dataset

model_name = "EleutherAI/gpt-neo-2.7B"

# 4-bit quantization config (QLoRA) — drop this and load in fp16/bf16 if you have enough VRAM
bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_use_double_quant=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype="bfloat16",
)

model = AutoModelForCausalLM.from_pretrained(
    model_name,
    quantization_config=bnb_config,
    device_map="auto",
)

qgfd_model = wrap_model_with_qgfd(
    model,
    MultiHeadQGFDLayer,
    diffusion_steps=2,
    target_alpha=0.05,
    warmup_steps=2000,
    kernel_size=5,
    early_stop_eps=0.0,
    verbose=True)

model = prepare_model_for_kbit_training(qgfd_model)

tokenizer = AutoTokenizer.from_pretrained(model_name)
if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

# LoRA config — GPT-Neo attention modules are named differently from Llama-style models
lora_config = LoraConfig(
    r=16,
    lora_alpha=32,
    lora_dropout=0.05,
    bias="none",
    task_type="CAUSAL_LM",
    target_modules=["k_proj", "v_proj", "q_proj", "out_proj"],  # GPT-Neo attention proj names
)

# --- Dataset ---
ds = load_dataset("arbml/alpagasus_cleaned")

# Check the actual columns first — alpaca-style sets usually have
# instruction / input / output. Adjust the formatting function to match.
print(ds["train"].column_names)
print(ds["train"][0])

def format_example(example):
    if example.get("input"):
        prompt = (
            f"### Instruction:\n{example['instruction']}\n\n"
            f"### Input:\n{example['input']}\n\n### Response:\n{example['output']}"
        )
    else:
        prompt = (
            f"### Instruction:\n{example['instruction']}\n\n### Response:\n{example['output']}"
        )
    return {"text": prompt}

train_ds = ds["train"].map(format_example)

# --- Training ---
training_args = SFTConfig(
    output_dir="./gpt-neo-2.7b-alpagasus-lora",
    per_device_train_batch_size=4,
    gradient_accumulation_steps=4,
    num_train_epochs=3,
    learning_rate=2e-4,
    bf16=True,
    logging_steps=10,
    save_strategy="epoch",
    dataset_text_field="text",
    packing=False,
)

trainer = SFTTrainer(
    model=model,
    train_dataset=train_ds,
    args=training_args,
    peft_config=lora_config,
)

trainer.train()
trainer.save_model("./gpt-neo-2.7b-alpagasus-lora-final")
```
